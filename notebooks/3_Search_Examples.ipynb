# 3. Search Examples
Query the neural search API and compare with keyword search.

```python
import requests
import pandas as pd
import numpy as np
from tabulate import tabulate
```

```python
API_URL = "http://localhost:8000/api/v1"

def neural_search(query, top_k=5, use_rerank=False):
    resp = requests.post(f"{API_URL}/search", json={"query": query, "top_k": top_k, "use_rerank": use_rerank})
    return resp.json()
```

```python
queries = [
    "transformer model for nlp",
    "how do computers learn from data",
    "comparing different architectures",
    "quantum computing",
]
```

```python
for query in queries:
    print(f"\n{'='*60}")
    print(f"Query: {query}")
    result = neural_search(query)
    print(f"Intent: {result['intent']} | Results: {result['total']}")
    print("-" * 60)
    for r in result['results'][:3]:
        score_str = f"{r['score']:.4f}" if r.get('score') else "N/A"
        print(f"  [{score_str}] {r['title']}")
```

```python
# Compare neural vs keyword search for a specific query
query = "machine learning"
neural = neural_search(query, use_rerank=True)

print("Neural Search Results:")
for r in neural['results'][:5]:
    print(f"  {r['score']:.4f} - {r['title']}")
```
