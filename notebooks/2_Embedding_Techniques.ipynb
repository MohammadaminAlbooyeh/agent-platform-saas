# 2. Embedding Techniques
Compare different embedding models and visualize document vectors.

```python
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv('../data/raw/documents.csv')
texts = (df['title'] + ' ' + df['content']).tolist()
print(f"{len(texts)} documents loaded")
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = model.encode(texts, normalize_embeddings=True, show_progress_bar=True)
print(f"Embedding shape: {embeddings.shape}")
```

```python
pca = PCA(n_components=2)
emb_2d = pca.fit_transform(embeddings)
plt.figure(figsize=(12, 8))
scatter = plt.scatter(emb_2d[:, 0], emb_2d[:, 1], c=range(len(df)), cmap='viridis', s=100)
for i, title in enumerate(df['title']):
    plt.annotate(title[:30], (emb_2d[i, 0], emb_2d[i, 1]), fontsize=8, alpha=0.8)
plt.colorbar(scatter, label='Document Index')
plt.title('Document Embeddings (PCA 2D Projection)')
plt.tight_layout()
plt.show()
```

```python
query = "deep learning for language understanding"
query_vec = model.encode(query, normalize_embeddings=True)
similarities = embeddings @ query_vec
top_indices = np.argsort(similarities)[::-1][:5]
print(f"Query: {query}\n")
for i, idx in enumerate(top_indices):
    print(f"{i+1}. [{similarities[idx]:.4f}] {df.iloc[idx]['title']}")
```
