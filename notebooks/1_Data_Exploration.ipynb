# 1. Data Exploration
Explore the sample document dataset used for neural search.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('../data/raw/documents.csv')
df.head()
```

```python
print(f"Shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
df.info()
```

```python
df['word_count'] = df['content'].str.split().str.len()
df['char_count'] = df['content'].str.len()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].hist(df['word_count'], bins=15, edgecolor='black')
axes[0].set_title('Word Count Distribution')
axes[0].set_xlabel('Words')
axes[1].hist(df['char_count'], bins=15, edgecolor='black', color='green')
axes[1].set_title('Char Count Distribution')
axes[1].set_xlabel('Characters')
plt.tight_layout()
plt.show()
```

```python
print("Top sources:")
print(df['source'].value_counts())
```

```python
from wordcloud import WordCloud
text = ' '.join(df['content'].fillna(''))
wordcloud = WordCloud(width=800, height=400, background_color='white').generate(text)
plt.figure(figsize=(14, 6))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Content Word Cloud')
plt.show()
```
