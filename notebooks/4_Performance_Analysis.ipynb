# 4. Performance Analysis
Benchmark latency and recall for the neural search pipeline.

```python
import requests
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
API_URL = "http://localhost:8000/api/v1"

test_queries = [
    "neural networks", "transformer", "machine learning",
    "quantum", "computer vision", "reinforcement learning",
    "nlp", "attention", "similarity search", "deep learning",
]
```

```python
def benchmark(top_k_values=[5, 10, 25, 50]):
    results = []
    for top_k in top_k_values:
        latencies = []
        for query in test_queries:
            start = time.perf_counter()
            resp = requests.post(f"{API_URL}/search", json={"query": query, "top_k": top_k, "use_rerank": False})
            elapsed = (time.perf_counter() - start) * 1000
            latencies.append(elapsed)
        results.append({
            "top_k": top_k,
            "mean_ms": np.mean(latencies),
            "std_ms": np.std(latencies),
            "min_ms": np.min(latencies),
            "max_ms": np.max(latencies),
        })
    return pd.DataFrame(results)
```

```python
print("Running benchmark...")
df_perf = benchmark([5, 10, 25, 50])
print(df_perf.to_string(index=False))
```

```python
plt.figure(figsize=(10, 5))
plt.errorbar(df_perf['top_k'], df_perf['mean_ms'], yerr=df_perf['std_ms'], marker='o', capsize=5)
plt.xlabel('top_k')
plt.ylabel('Latency (ms)')
plt.title('Search Latency vs top_k')
plt.grid(True, alpha=0.3)
plt.show()
```
